# Gráfico de linha

```python
import matplotlib.pyplot as plt
import numpy as np
from cycler import cycler

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
vendas = [85, 90, 78, 92, 88, 96, 94, 98, 100, 102, 105, 108]

#paleta de cores
cores = plt.get_cmap('Set1').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle = ciclo_cores)

fig,ax = plt.subplots(figsize=(20,10))
ax.plot(meses, vendas)

#gráfico
marcadores = np.where(vendas == np.min(vendas), 'x', 'o')

for posicao, marcador in enumerate(marcadores):
    if marcador == 'x':
        cor = 'red'
    else:
        cor = 'red'
    ax.scatter(meses[posicao], vendas[posicao], marker=marcador, color=cor)

ax.annotate('Menor valor', color = 'red', xy = (2, 78), xytext = (2.2, 79))

#destaque
ax.axvspan(6, 11, color = cores[2], alpha = 0.2)

ax.set_xlabel('Mês')
ax.set_ylabel('Vendas (em milhares)')
ax.set_title('Crescimento das Vendas ao Longo do Ano')
#ax.grid(True)
plt.show()
```

# Gráfico de barras

```python
produtos = ['Produto A', 'Produto B', 'Produto C', 'Produto D', 'Produto E']
quantidades = [23, 45, 17, 30, 37]
dados_ordenados = sorted(zip(quantidades, produtos))
quantidades_ordenadas = [x[0] for x in dados_ordenados]
produtos_ordenados = [x[1] for x in dados_ordenados]

fig, ax = plt.subplots(figsize=(20,10))

barras = ax.barh(produtos_ordenados, quantidades_ordenadas)

ax.set_title('Quantidade de produtos vendidos')
ax.set_ylabel('Produtos')
ax.set_xlabel('Quantidade vendida')

ax.bar_label (barras, labels=[max(quantidades_ordenadas) if x == max(quantidades_ordenadas) else '' for x in quantidades_ordenadas], padding = 3, fontsize = 10, fontweight = 'bold')

# Retirando a borda
ax.set_frame_on(False)

# Removendo ticks
ax.set_xticks([])
ax.tick_params(axis='y', length = 0)

# Core das barras
for i, barra in enumerate(barras):

    if i != quantidades_ordenadas.index(max(quantidades_ordenadas)):
        barras[i].set_color(cores[0])
        barras[i].set_alpha(0.4)


plt.show
```

# Gráfio de pizza

```python
categorias = ['Eletrônicos', 'Roupas', 'Alimentos', 'Livros']
quantidades_vendidas = [45, 30, 15, 10]
destaque = [0.1, 0, 0, 0]  # destacar a primeira fatia

fig, ax = plt.subplots()
plt.style.use('seaborn-v0_8-dark-palette')
ax.pie(quantidades_vendidas, labels=categorias, explode=destaque, autopct='%.1f')
ax.set_title('Categorias de produtos vendidos')

plt.show()
```

# Gráfico de barras empilhadas

```python
anos = [2018, 2019, 2020, 2021]
vendas_online = [5, 10, 15, 20]
vendas_loja = [10, 8, 6, 4]

# rc = runtime configuration (configuração quando executa)
cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

fig, ax = plt.subplots()


barras_online = ax.bar(anos, vendas_online, label='Online')
barras_loja = ax.bar(anos, vendas_loja, bottom=vendas_online, label='Loja')

ax.bar_label(barras_online, labels = vendas_online, label_type = 'center', fontweight='bold')
ax.bar_label(barras_loja, labels = vendas_loja, label_type = 'center', fontweight='bold')

# forçar os anos a serem exibidos como inteiros
ax.set_xticks(ticks=anos)

# Desativando Caixa
ax.set_frame_on(False)

ax.set_title('Vendas Online X Loja Física (em milhões)')
ax.legend()
ax.tick_params(axis = 'x', length=0)
#ax.set_yticks([])
plt.show()
```
