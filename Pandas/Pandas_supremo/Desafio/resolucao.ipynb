```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
vendas_df = pd.read_csv("Vendas_Chocolate.csv", sep=';')
display(vendas_df.info())
```

Vamos tratar a tabela primeiro

```python
# Valores vazios

vendas_df = vendas_df.drop(columns="Data de Hoje") # coluna que tem muitos valores vazios e vai atrapalhar a analise. está sendo removida do df 
vendas_df = vendas_df.dropna() # excluindo linhas que tem algum valor vazaio! somente foi feito por que não influeciaram na analise.

# Valores com formatação errado

# data
vendas_df["Data"] = pd.to_datetime(vendas_df["Data"], dayfirst=True)

# valor
vendas_df["Valor"] = vendas_df["Valor"].str.replace('$','').str.replace(',','').astype(float)

# Valores duplicados
vendas_df = vendas_df.drop_duplicates()

vendas_df = vendas_df.reset_index(drop=True)
display(vendas_df.info())
```

```python
display(vendas_df)
```

```python
vendas_df["Ano"] = vendas_df["Data"].dt.year
vendas_df['Mes'] = vendas_df['Data'].dt.to_period('M')
```

Analise de crescimento

```python
vendas_mes = vendas_df[['Mes', 'Valor']].sort_values(by='Mes').groupby('Mes', as_index=False).sum()
display(vendas_mes)

grafico = px.bar(vendas_mes, x=vendas_mes['Mes'].astype(str), y='Valor')#por uma configuração do plotly, não é possível usar o formato de data da coluna mes, portanto devemos usar essa configuração para indicar que ela deve ser lida com string
grafico.show()
```

Analise de ticket médio

```python
vendas_anuais = vendas_df[["Ano", "Valor", 'Caixas Enviadas']].groupby('Ano', as_index=False).sum()
vendas_anuais['Ticket Medio'] = vendas_anuais['Valor'] / vendas_anuais['Caixas Enviadas']
vendas_anuais['% Crescimento Valor'] = vendas_anuais['Valor'].pct_change()
display(vendas_anuais.style.format({
    "Valor": "${:,.2f}",
    "Caixas Enviadas": "{:,.2f}",
    "Ticket Medio": "${:,.2f}",
    '% Crescimento Valor' : '{:,.1%}'
}))
```

Analise por produto

```python
df_produtos_anual = vendas_df[["Ano", "Produto", "Valor"]]

df_produtos_anual = df_produtos_anual.groupby(["Ano", "Produto"], as_index=False).sum()

grafico = px.bar(df_produtos_anual, x="Ano", y="Valor", color='Produto')
grafico.show()
```
