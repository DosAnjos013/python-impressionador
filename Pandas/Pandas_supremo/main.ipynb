```python
# Removendo colunas desnecessárias
df_vendas_format = df_vendas.drop(columns=['Data_Base'])

# Completando espaços vazios com o valor que não foi preenchido por erro operacional
df_vendas_format['Loja'] = df_vendas_format['Loja'].fillna('Online')

# Corrigindo o formato dos dados
df_vendas_format['Data'] = pd.to_datetime(df_vendas_format['Data'], format="%Y-%m-%d")

# Padronizando textos
df_vendas_format['Loja'] = df_vendas_format['Loja'].str.strip().str.title()

# Removendo duplicatas
df_vendas_format = df_vendas_format.drop_duplicates(subset=['ID_Pedido'])


display(df_vendas_format.sample(15))
display(df_vendas_format.info())
```

```python
# Criar coluna de faturamento

df_vendas_format["Faturamento"] = df_vendas_format['Qtd'] * df_vendas_format["Preco_Unitario"]

# Criar coluna de forma de vendas

df_vendas_format["Forma_venda"] = np.where(df_vendas_format["Loja"] == "Online", "Online", "Presencial")

# Criar coluna que informa a região de vendas


dic_regioes = {
    'São Paulo' : 'Sudeste',
    'Belo Horizonte' : 'Sudeste',
    'Online' : 'Online',
    'Rio De Janeiro' : 'Sudeste',
    'Salvador' : 'Nordeste',
    'Recife' : 'Nordeste',
    'Curitiba' : 'Sul',
    'Porto Alegre' : 'Sul'
}

df_vendas_format['Região'] = df_vendas_format['Loja'].map(dic_regioes)
```
