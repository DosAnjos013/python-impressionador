Contabilizar a origem de cada aluno

```python
with open('Alunos.txt', 'r') as arquivo:
    linhas = arquivo.readlines()
    del linhas[:4]

    qtd_anuncio = 0
    qtd_org = 0
    qtd_yt_org = 0
    qtd_igfb_org = 0
    qtd_site_org = 0 

    for linha in linhas:
        email, origem = linha.split(',')

        if 'org' in origem:
            qtd_org += 1
            if 'yt_org' in origem:
                qtd_yt_org += 1
            if 'site_org' in origem:
                qtd_site_org += 1
            if 'ig_org' in origem or 'igfb_org' in origem:
                qtd_igfb_org += 1
        else:
            qtd_anuncio += 1

print(f'Quantidade anuncio = {qtd_anuncio}\nQuantidade orgânico = {qtd_org}\nQuantidade orgânica do youtube = {qtd_yt_org}\nQuantidade orgânica do site = {qtd_site_org}\nQuantidade orgânica das redes sociais = {qtd_igfb_org}')

with open('Desafio.txt', 'w') as arquivo:
    arquivo.write('Quantidade anuncio = {}\n'.format(qtd_anuncio))
    arquivo.write('Quantidade orgânico = {}\n'.format(qtd_org))
    arquivo.write('Quantidade orgânica do youtube = {}\n'.format(qtd_yt_org))
    arquivo.write('Quantidade orgânica do site = {}\n'.format(qtd_site_org))
    arquivo.write('Quantidade orgânica das redes sociais = {}\n'.format(qtd_igfb_org))
```
