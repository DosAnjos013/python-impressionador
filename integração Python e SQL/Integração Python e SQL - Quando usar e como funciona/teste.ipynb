### Para fazer a conexão, vamos precisar preencher 3 parametros:

1 - 'Driver' = {Qual o driver a ser usado. ("fornecedor" do banco de dados)} (atenção as chaves nesse parametro);

2 - 'Server' = Onde está o banco de dados (link ou local);

3 - 'Database' = Nome do banco de dados.

Exp 1:
conexao = ('Driver={SQL Server};Server=localhost;Database=exemplo.db')

Exp 2:
conexao = ('Driver={SQL Server};\n''Server=localhost;\n''Database=exemplo.db')

```python
dados_conexao = ('Driver={SQLite3 ODBC Driver};Server=localhost;Database=salarios.sqlite')

'''
Se o banco de dados tiver dados de autenticação, você pode adicionar o usuário e a senha na string de conexão:
dados_conexao = ('Driver={SQLite3 ODBC Driver};Server=localhost;Database=chinook.db;UID=usuario;PWD=senha')
'''

conexao = pyodbc.connect(dados_conexao)

cursor = conexao.cursor()
```

```python
''' Qualquer alteração será feita através do comando execute, que recebe como parâmetro a query SQL que será executada no banco de dados. Por exemplo, para criar uma tabela chamada "clientes" com os campos "id", "nome" e "email", podemos fazer o seguinte:
cursor.execute("")'''

cursor.execute("SELECT * FROM Salaries")
# Para pode consultar valores do banco de dados:
valores = cursor.fetchall()
```

```python
display(valores[:10])

```

```python
cursor.close()
conexao.close() # finalizando a conexão com o banco de dados
```
